```sql
SELECT    DISTINCT(`Sample ID`),
         `Instrument Serial No`,
        `Run Date`, 
      `CD4+ Count/uL`,
      `CD4+ Lymphs Percent`,
      `**CD45+ Count/uL`
FROM downtime.aquios_downtime.downtime_1
WHERE `Sample ID`  LIKE '%P' OR
        `Sample ID`  LIKE '%NOF'OR
         `Sample ID`  LIKE 'HNM%'
UNION

SELECT    DISTINCT(`Sample ID`),
         `Instrument Serial No`,
        `Run Date`, 
      `CD4+ Count/uL`,
      `CD4+ Lymphs Percent`,
      `**CD45+ Count/uL`
FROM downtime.aquios_downtime.downtime_2
WHERE `Sample ID`  LIKE '%P' OR
        `Sample ID`  LIKE '%NOF'OR
         `Sample ID`  LIKE 'HNM%'
ORDER BY `Run Date`;


SELECT 
        COUNT(DISTINCT `Sample ID`)
      
FROM downtime.aquios_downtime.downtime_1
WHERE `Sample ID` LIKE '%P' OR
        `Sample ID` LIKE '%NOF'OR
         `Sample ID` LIKE 'HNM%'
UNION
SELECT 
        COUNT(DISTINCT `Sample ID`)
      
FROM downtime.aquios_downtime.downtime_2
WHERE `Sample ID` LIKE '%P' OR
        `Sample ID` LIKE '%NOF'OR
         `Sample ID` LIKE 'HNM%';

         

SELECT `Sample ID`,
        `Instrument Serial No`,
        `Run Date`, 
        MIN(`CD4+ Count/uL`),
        `CD4+ Lymphs Percent`
     -- `**CD45+ Count/uL`
FROM downtime.aquios_downtime.downtime_1
GROUP BY `Sample ID`, `Instrument Serial No`, `Run Date`, `CD4+ Lymphs Percent`
UNION
SELECT  `Sample ID`,
        `Instrument Serial No`,
        `Run Date`, 
        MIN(`CD4+ Count/uL`),
        `CD4+ Lymphs Percent`
      --  `**CD45+ Count/uL`     
FROM downtime.aquios_downtime.downtime_2
GROUP BY `Sample ID`, `Instrument Serial No`, `Run Date`, `CD4+ Lymphs Percent`
ORDER BY `Run Date`;

ALTER TABLE downtime.aquios_downtime.downtime_2
ALTER COLUMN `CD4+ Count/uL`
TYPE BIGINT;


SELECT   *   
FROM downtime.aquios_downtime.downtime_1
UNION
SELECT *
FROM downtime.aquios_downtime.downtime_2;

SELECT  `Run Date`,
        --`Notification Codes`,
       -- `Flag Codes`,
        `Sample ID`,
        WCC,
        `Lymphs Percent`,
        `CD4+ Count/uL`,
        CASE 
                WHEN `CD4+ Count/uL` < 100 THEN 'CRAG TEST'
                ELSE ''
        END AS `CRAG TEST`

FROM downtime.aquios_downtime.downtimedata
WHERE `Sample ID` NOT LIKE '%P' OR
        `Sample ID` NOT LIKE '%NOF'OR
         `Sample ID` NOT LIKE 'HNM%' OR
         `Sample ID` NOT LIKE 'IMMU%';


SELECT DISTINCT `Flag Codes`,
        `Notification Codes`
FROM downtime.aquios_downtime.downtimedata;

UPDATE downtime.aquios_downtime.downtimedata
SET `Sample ID` = 'IMMUNOTROL NORMAL'
WHERE `Sample ID` = '6300195';
```
