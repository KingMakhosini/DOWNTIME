```sql
SELECT  `Run Date`,
        `Notification Codes`,
       -`Flag Codes`,
        `Sample ID`,
        WCC,
        `Lymphs Percent`,
        `CD4+ Count/uL`,
        CASE 
                WHEN `CD4+ Count/uL` < 100 THEN 'CRAG TEST'
                ELSE ''
        END AS `CRAG TEST`

FROM downtime.aquios_downtime.downtimedata
WHERE `Sample ID`  LIKE '%P' OR
        `Sample ID`  LIKE '%NOF'OR
         `Sample ID`  LIKE 'HNM%' OR
         `Sample ID` NOT LIKE 'IMMU%';
```
